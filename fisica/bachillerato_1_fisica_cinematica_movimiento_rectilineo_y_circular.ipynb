## 1. Conceptos y lenguaje

- **Posición** $x$ (o $\vec r$): lugar del móvil respecto a un origen.
- **Desplazamiento** $\Delta x=x_f-x_i$: cambio de posición, con signo. En varias dimensiones: $\Delta\vec r=\vec r_f-\vec r_i$.
- **Distancia recorrida**: longitud total del camino; siempre no negativa. No tiene por qué coincidir con $|\Delta x|$.
- **Velocidad media**: $v_m=\Delta x/\Delta t$ (en una dimensión). **Rapidez media**: distancia total dividida entre tiempo total.
- **Velocidad instantánea** $v=dx/dt$: indica rapidez y sentido. **Aceleración** $a=dv/dt$: cambio de velocidad por unidad de tiempo.

Unidades: posición y distancia en **m**, tiempo en **s**, velocidad en **m/s**, aceleración en **m/s²**.

**Conversión:** de km/h a m/s se divide entre 3,6; de m/s a km/h se multiplica por 3,6. $1\ \mathrm{rpm}=2\pi/60\ \mathrm{rad/s}$.

> **Truco:** antes de calcular, dibuja el eje positivo. El signo de la velocidad dice hacia dónde se mueve; el signo de la aceleración dice hacia dónde cambia la velocidad. Aceleración negativa no significa necesariamente frenar.

## 2. MRU: movimiento rectilíneo uniforme

Trayectoria recta, velocidad constante, aceleración nula:

$$\boxed{x(t)=x_0+vt},\qquad \boxed{a=0}$$

La gráfica $x$–$t$ es una recta cuya pendiente es $v$; la gráfica $v$–$t$ es horizontal.

**Ejemplo.** Un ciclista parte de $x_0=20\,\mathrm m$ y avanza a $5\,\mathrm{m/s}$. Tras 12 s: $x=20+5\cdot12=80\,\mathrm m$; desplazamiento $60\,\mathrm m$.

**Encuentros:** escribe una función de posición para cada móvil usando **el mismo reloj** y resuelve $x_1(t)=x_2(t)$. Comprueba que el instante encontrado está dentro del intervalo en el que ambas ecuaciones son válidas.

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 12, 200)
plt.plot(t, 20 + 5*t, label='x(t) = 20 + 5t')
plt.xlabel('Tiempo (s)'); plt.ylabel('Posición (m)')
plt.title('MRU: posición frente al tiempo'); plt.grid(True); plt.legend(); plt.show()
```

## 3. MRUA: movimiento rectilíneo uniformemente acelerado

Aceleración constante en una trayectoria recta:

$$\boxed{v=v_0+at}$$
$$\boxed{x=x_0+v_0t+\tfrac12at^2}$$
$$\boxed{v^2=v_0^2+2a(x-x_0)}$$

La última ecuación elimina el tiempo y es útil cuando no lo conocemos. La pendiente de $v$–$t$ es la aceleración; el área **con signo** bajo la gráfica $v$–$t$ es el desplazamiento.

**Ejemplo.** Un coche lleva $v_0=20\,\mathrm{m/s}$ y frena con $a=-4\,\mathrm{m/s^2}$. Se detiene cuando $0=20-4t$, es decir, $t=5\,\mathrm s$. Recorre $\Delta x=20(5)-\tfrac12(4)(5^2)=50\,\mathrm m$.

**Ojo:** si $v$ y $a$ tienen el mismo signo, aumenta la rapidez; si tienen signos contrarios, disminuye.

```python
t = np.linspace(0, 6, 200)
v = 20 - 4*t
plt.plot(t, v); plt.axhline(0, linewidth=0.8)
plt.axvline(5, linestyle='--', label='Se detiene: 5 s')
plt.xlabel('Tiempo (s)'); plt.ylabel('Velocidad (m/s)')
plt.title('Frenado con aceleración constante'); plt.grid(True); plt.legend(); plt.show()
```

## 4. Caída libre y lanzamiento vertical

Es un MRUA en el eje vertical, **si despreciamos la resistencia del aire**. Si elegimos **arriba como positivo**:

$$\boxed{y=y_0+v_0t-\tfrac12gt^2},\quad \boxed{v=v_0-gt},\quad g\approx9{,}8\,\mathrm{m/s^2}$$

- **Subida:** $v>0$, aunque la aceleración es $-g$.
- **Altura máxima:** $v=0$ en ese instante; **la aceleración sigue siendo $-g$**.
- **Bajada:** $v<0$.
- **Misma altura de salida y llegada:** sin rozamiento, la rapidez al volver es igual a la inicial, pero la velocidad tiene signo opuesto: $v_{\rm vuelta}=-v_0$.
- Si se elige **abajo positivo**, la aceleración es $+g$ y deben cambiarse coherentemente los signos.

**Ejemplo.** Se lanza una pelota hacia arriba a $19{,}6\,\mathrm{m/s}$ desde el suelo. Alcanza su altura máxima en $t=19{,}6/9{,}8=2\,\mathrm s$ y sube $h=19{,}6(2)-4{,}9(2^2)=19{,}6\,\mathrm m$. Vuelve al suelo en $4\,\mathrm s$ con $v=-19{,}6\,\mathrm{m/s}$.

> **Importante:** igualdad de rapidez al subir y bajar solo al comparar **la misma altura**, bajo gravedad constante y sin rozamiento; no vale comparar instantes arbitrarios.

```python
t = np.linspace(0, 4, 200)
y = 19.6*t - 4.9*t**2
v = 19.6 - 9.8*t
plt.plot(t, y, label='Altura y(t)')
plt.scatter([2], [19.6], label='Altura máxima')
plt.xlabel('Tiempo (s)'); plt.ylabel('Altura (m)')
plt.title('Lanzamiento vertical'); plt.grid(True); plt.legend(); plt.show()
```

## 5. Dos móviles y distintos instantes de inicio

**Una sola línea temporal:** si el móvil A empieza en $t=0$ y B empieza en $t=15\,\mathrm s$, el tiempo transcurrido para B es $t_B=t-15$, **solo para $t\ge15$**.

**Ejemplo.** A sale de $x=0$ a $20\,\mathrm{m/s}$ en $t=0$. B sale del mismo lugar 15 s después, a velocidad constante $v_B$. Si B alcanza a A en el instante global $t=90\,\mathrm s$:

$$x_A=20t,\qquad x_B=v_B(t-15).$$

Al igualar en $t=90$: $20\cdot90=v_B(90-15)$, luego $\boxed{v_B=24\,\mathrm{m/s}}$.

**Encuentros verticales:** para dos objetos, escribe $y_1(t)$ y $y_2(t)$ con un eje común, y resuelve $y_1=y_2$. Si ambos tienen la misma aceleración gravitatoria, los términos $-\tfrac12gt^2$ se cancelan cuando usan el mismo tiempo global. Después comprueba si cada uno **sube o baja** evaluando $v_i(t)$ en el instante encontrado: no hay que adivinarlo de antemano.

**Ejemplo.** Desde 20 m se lanza una piedra hacia arriba a 25 m/s y, simultáneamente, desde el suelo otra a 30 m/s, ambas bajo la misma gravedad. $y_1=20+25t-4{,}9t^2$ y $y_2=30t-4{,}9t^2$. Se encuentran a $t=4\,\mathrm s$ y $y=41{,}6\,\mathrm m$. Sus velocidades son $-14{,}2$ y $-9{,}2\,\mathrm{m/s}$: **las dos bajan**.

## 6. Movimiento circular: magnitudes angulares

En una circunferencia de radio $R$, el ángulo $\varphi$ se mide en **radianes**: una vuelta son $2\pi\,\mathrm{rad}$. El arco recorrido es $s=R\,\Delta\varphi$ (si $\Delta\varphi$ es el ángulo barrido, con signo; la distancia sobre el arco usa su valor absoluto cuando no cambia el sentido).

$$\boxed{\omega=\frac{d\varphi}{dt}},\qquad \boxed{\alpha=\frac{d\omega}{dt}}$$

$$\boxed{v=|\omega|R},\qquad \boxed{a_t=\alpha R},\qquad \boxed{a_n=\omega^2R=\frac{v^2}{R}}$$

$a_t$ es la **componente tangencial con signo**; $a_n$ es el **módulo** de la aceleración normal, siempre dirigido hacia el centro. La aceleración total es vectorial: tangencial y normal son perpendiculares; su módulo es $\sqrt{a_t^2+a_n^2}$.

**Periodo** $T$: segundos por vuelta. **Frecuencia** $f$: vueltas por segundo, en Hz.

$$\boxed{f=1/T},\qquad \boxed{|\omega|=2\pi/T=2\pi f}$$

> No confundir $\omega$ (rad/s) con $f$ (Hz): una vuelta equivale a $2\pi$ radianes.

## 7. MCU: movimiento circular uniforme

El radio es constante y la velocidad angular es constante: $\alpha=0$.

$$\boxed{\varphi(t)=\varphi_0+\omega t}$$

Aunque el módulo de la velocidad tangencial es constante, **su dirección cambia**, por lo que hay aceleración normal $a_n=\omega^2R$; la aceleración tangencial es cero.

**Ejemplo.** Una rueda de radio $0{,}25\,\mathrm m$ gira a 360 rpm. $\omega=360\cdot2\pi/60=12\pi\approx37{,}70\,\mathrm{rad/s}$. Entonces $v=\omega R\approx9{,}42\,\mathrm{m/s}$, $T=2\pi/\omega\approx0{,}167\,\mathrm s$ y $a_n=\omega^2R\approx355{,}3\,\mathrm{m/s^2}$.

**Posición en el plano:** $x=R\cos\varphi$, $y=R\sin\varphi$, tomando el centro como origen.

```python
R = 1
phi = np.linspace(0, 2*np.pi, 400)
fig, ax = plt.subplots()
ax.plot(R*np.cos(phi), R*np.sin(phi))
a = np.pi/4
x, y = np.cos(a), np.sin(a)
ax.scatter([x], [y]); ax.arrow(x, y, -0.45*x, -0.45*y, head_width=.06, length_includes_head=True)
ax.arrow(x, y, -0.4*y, 0.4*x, head_width=.06, length_includes_head=True)
ax.text(x-0.35*x, y-0.35*y, 'a normal')
ax.text(x-0.5*y, y+0.5*x, 'v tangencial')
ax.set_aspect('equal'); ax.grid(True); ax.set_title('MCU: velocidad tangente y aceleración hacia el centro'); plt.show()
```

## 8. MCUA: movimiento circular uniformemente acelerado

La aceleración angular $\alpha$ es constante. Es el equivalente angular del MRUA:

$$\boxed{\omega=\omega_0+\alpha t}$$
$$\boxed{\varphi=\varphi_0+\omega_0t+\tfrac12\alpha t^2}$$
$$\boxed{\omega^2=\omega_0^2+2\alpha(\varphi-\varphi_0)}$$

**Aceleraciones:** $a_t=\alpha R$ es constante, mientras que $a_n=\omega(t)^2R$ **varía con el tiempo**.

**Ejemplo.** Una rueda parte del reposo, $R=0{,}25\,\mathrm m$ y $\alpha=3{,}77\,\mathrm{rad/s^2}$. Tras 5 s: $\omega=18{,}85\,\mathrm{rad/s}$, $v=4{,}71\,\mathrm{m/s}$, $a_t=0{,}943\,\mathrm{m/s^2}$ y $a_n\approx88{,}8\,\mathrm{m/s^2}$.

> **Corrección importante frente a los apuntes:** al calcular $a_n$ hay que elevar al cuadrado la velocidad angular **del instante considerado**; las unidades del resultado son m/s². No debe confundirse $a_n$ con $a_t$.

```python
t = np.linspace(0, 5, 200)
w = 3.77*t
plt.plot(t, .25*3.77*np.ones_like(t), label='a tangencial')
plt.plot(t, .25*w*w, label='a normal')
plt.xlabel('Tiempo (s)'); plt.ylabel('Aceleración (m/s²)')
plt.title('MCUA: aceleraciones tangencial y normal'); plt.grid(True); plt.legend(); plt.show()
```
