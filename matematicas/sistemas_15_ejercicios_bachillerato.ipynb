### 9. Problema 12 · pág. 10 — Vehículo todoterreno

Un vehículo todoterreno sube las cuestas a **54 km/h**, las baja a **90 km/h** y en llano lleva una velocidad de **80 km/h**. Para ir de una ciudad a otra, separadas por **201 km**, emplea **2,8 horas** y para volver **2,6 horas**. ¿Cuántos kilómetros de camino llano hay entre ambas ciudades?

### 10. Problema 21 · pág. 11 — Videoclub

Un videoclub está especializado en películas de tres tipos: **infantiles, oeste americano y terror**. Se sabe que el **60 %** de las películas infantiles más el **50 %** de las del oeste representan el **30 %** del total de películas. El **20 %** de las infantiles más el **60 %** de las del oeste más el **60 %** de las de terror representan la **mitad** del total de películas. Hay **100 películas más del oeste que infantiles**. Halla el número de películas de cada tipo.

### 11. Problema 7 · pág. 9 — Venta de videojuegos

Una tienda ha vendido **600 ejemplares** de un videojuego por un total de **6384 €**. El original costaba **12 €**, pero también ha vendido copias con descuentos del **30 %** y del **40 %**. Sabiendo que el número total de copias vendidas fue **la mitad del número de originales**, calcula **cuántas copias se vendieron con un descuento del 30 %**.


### 12. Problema 8 · pág. 9 — Número de tres cifras

Halla un **número de tres cifras** sabiendo que la suma de sus cifras es **9**; que al restarle el número obtenido al **invertir el orden de sus cifras** se obtiene **198**; y que la cifra de las **decenas es la media aritmética** de las otras dos.


### 13. Problema 14 · págs. 10–11 — Espectadores de cine

Un cine proyecta una película únicamente **lunes, martes y miércoles**. El martes hubo un **12 % más** de espectadores que el lunes; el miércoles hubo un **12 % menos** que el martes; y el lunes asistieron **36 espectadores más** que el miércoles. ¿Cuántos espectadores hubo cada día?


### 14. Problema 19 · pág. 11 — Aleación de monedas

Un joyero tiene tres clases de monedas, **A, B y C**. Cada moneda A contiene **2 g de oro, 4 g de plata y 14 g de cobre**; cada moneda B, **6 g de oro, 4 g de plata y 10 g de cobre**; y cada moneda C, **8 g de oro, 6 g de plata y 6 g de cobre**. ¿Cuántas monedas de cada tipo debe fundir para obtener **44 g de oro, 44 g de plata y 112 g de cobre**?


### 15. Problema 20 · pág. 11 — Distribución de electrodomésticos

Un fabricante produce **42 electrodomésticos** que se distribuyen entre tres tiendas. La **primera** solicita tantas unidades como la segunda y la tercera **juntas**. La **segunda** solicita un **20 % más que la suma** de la mitad de lo pedido por la primera y la tercera parte de lo pedido por la tercera. ¿Cuántas unidades solicita cada tienda?
