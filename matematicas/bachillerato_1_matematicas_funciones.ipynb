# Funciones: dominio, composición e inversa
### Matemáticas · 1.º de Bachillerato

**Objetivos:** comprender qué valores puede tomar la variable, cómo encadenar funciones y cuándo una función puede deshacerse mediante otra. Las gráficas se generan al ejecutar las celdas de Python.

**Cómo utilizar el notebook:** ejecuta las celdas en orden. Modifica las funciones de los ejemplos para experimentar. Se emplean números reales y ángulos en **radianes**.

```python
import numpy as np
import matplotlib.pyplot as plt

def dibujar(curvas, xlim=(-6, 6), ylim=(-6, 6), titulo='', diagonal=False):
    fig, ax = plt.subplots(figsize=(8, 4.7))
    for x, y, etiqueta in curvas:
        ax.plot(x, y, lw=2, label=etiqueta)
    if diagonal:
        t = np.linspace(max(xlim[0], ylim[0]), min(xlim[1], ylim[1]), 200)
        ax.plot(t, t, '--', color='gray', label='y = x')
    ax.axhline(0, color='black', lw=.7); ax.axvline(0, color='black', lw=.7)
    ax.set(xlim=xlim, ylim=ylim, title=titulo, xlabel='x', ylabel='y')
    ax.grid(alpha=.25); ax.legend(); plt.show()

x = np.linspace(-6, 6, 1200)
```

## 1. Dominio y recorrido

Una función asigna a cada entrada $x$ de su **dominio** un único resultado $f(x)$. El **recorrido** (o imagen) es el conjunto de resultados que realmente obtiene.

- **Dominio:** valores de entrada para los que la expresión tiene sentido.
- **Recorrido:** valores de salida que alcanza la función.
- **Interpretación gráfica:** el dominio se lee sobre el eje horizontal; el recorrido, sobre el vertical.

**Notación:** $[a,b]$ incluye extremos; $(a,b)$ los excluye. Una unión $A\cup B$ reúne valores; una intersección $A\cap B$ exige cumplir ambas condiciones.

Ejemplo: $f(x)=x^2$ tiene dominio $\mathbb R$ y recorrido $[0,+\infty)$.

```python
dibujar([(x, x**2, 'f(x) = x²')], ylim=(-1, 12), titulo='Dominio: todos los reales; recorrido: y ≥ 0')
```

## 2. Reglas de dominio

| Tipo | Condición | Ejemplo |
|---|---|---|
| Polinómica | Ninguna: $D=\mathbb R$ | $x^2-3x$ |
| Racional | Denominador $\ne 0$ | $1/(x-2)$ |
| Raíz **par** | Radicando $\ge 0$ | $\sqrt{x+2}$ |
| Raíz **impar** | El radical no impone restricciones | $\sqrt[3]{x-2}$ |
| Logaritmo | Argumento $>0$ | $\ln(x-1)$ |
| Exponencial | $D=\mathbb R$ si la base es positiva | $2^x$ |
| Seno y coseno | $D=\mathbb R$ | $\sin x$ |
| Tangente | $\cos x\ne0$ | $\tan x$ |

**¿Por qué las raíces impares admiten negativos?** Porque $(-2)^3=-8$, de modo que $\sqrt[3]{-8}=-2$. En cambio, ningún real elevado al cuadrado da $-8$.

**Atención:** si una raíz está en un denominador, además debe ser distinta de cero. Así, $1/\sqrt{x-2}$ exige $x>2$ y $1/\sqrt[3]{x-2}$ exige $x\ne2$.

```python
xp = np.linspace(-1.95, 6, 700)
xc = np.linspace(-6, 6, 900)
dibujar([(xp, np.sqrt(xp+2), '√(x+2)'), (xc, np.cbrt(xc+2), '∛(x+2)')], ylim=(-3, 4), titulo='Raíz par frente a raíz impar')
```

```python
a = np.linspace(-6, 1.96, 500); c = np.linspace(2.04, 6, 500)
dibujar([(a, 1/(a-2), '1/(x−2), izquierda'), (c, 1/(c-2), '1/(x−2), derecha')], ylim=(-6, 6), titulo='La función racional no está definida en x = 2')
```

```python
a = np.linspace(1.01, 7, 600)
dibujar([(a, np.log(a-1), 'ln(x−1)'), (x, 2.0**x, '2ˣ')], xlim=(-4, 7), ylim=(-5, 8), titulo='Logaritmo: x > 1; exponencial: todo R')
```

### 2.1 Inecuaciones y tabla de signos

Para $f(x)=\sqrt{x^2-5x+6}$ necesitamos

$$x^2-5x+6=(x-2)(x-3)\ge 0.$$

Las raíces dividen la recta en tres intervalos. El producto es positivo antes de $2$, negativo entre $2$ y $3$, y positivo después de $3$.

$$\boxed{D=(-\infty,2]\cup[3,+\infty)}$$

**No basta con encontrar las raíces:** hay que determinar en qué intervalos se cumple la desigualdad.

```python
a=np.linspace(-2,2,400); c=np.linspace(3,7,400)
dibujar([(a,np.sqrt(np.maximum((a-2)*(a-3),0)),'x ≤ 2'),(c,np.sqrt(np.maximum((c-2)*(c-3),0)),'x ≥ 3')], xlim=(-2,7), ylim=(-.5,7), titulo='√((x−2)(x−3)): hueco entre 2 y 3')
```

### 2.2 Varias restricciones a la vez

Para $f(x)=\dfrac{\sqrt{x+2}}{x-3}$:

1. Raíz par: $x+2\ge0\Rightarrow x\ge-2$.
2. Denominador: $x-3\ne0\Rightarrow x\ne3$.
3. **Intersección:** $\boxed{D=[-2,3)\cup(3,+\infty)}$.

**Caso especial:** simplificar una fracción no devuelve al dominio los puntos originalmente prohibidos. Por ejemplo, $(x^2-1)/(x-1)=x+1$ solo cuando $x\ne1$.

### 2.3 Trigonometría

$\sin x$ y $\cos x$ están definidas para todo real. Como $\tan x=\sin x/\cos x$, excluye $x=\pi/2+k\pi$, para cualquier entero $k$.

```python
t=np.linspace(-2*np.pi,2*np.pi,1600); tan=np.tan(t); tan[np.abs(np.cos(t))<.06]=np.nan
dibujar([(t,np.sin(t),'sen(x)'),(t,np.cos(t),'cos(x)'),(t,tan,'tan(x)')], xlim=(-2*np.pi,2*np.pi), ylim=(-4,4), titulo='Trigonometría: tangente con discontinuidades')
```

## 3. Composición de funciones

**Componer** consiste en aplicar una función al resultado de otra:

$$\boxed{(f\circ g)(x)=f(g(x))}$$

Se lee «$f$ compuesta con $g$»: **primero $g$ y después $f$**.

Si $g(x)=x+1$ y $f(x)=x^2$, entonces:

$$f(g(x))=(x+1)^2,\qquad g(f(x))=x^2+1.$$

**¡El orden importa!** En general $f\circ g\ne g\circ f$.

```python
dibujar([(x,(x+1)**2,'f∘g = (x+1)²'),(x,x**2+1,'g∘f = x²+1')], xlim=(-4,3), ylim=(-1,12), titulo='La composición no suele ser conmutativa')
```

### 3.1 Caso general: composición de $n$ funciones

Para funciones $f_1,f_2,\ldots,f_n$, definimos

$$F=(f_n\circ f_{n-1}\circ\cdots\circ f_2\circ f_1),$$

es decir,

$$\boxed{F(x)=f_n(f_{n-1}(\cdots f_2(f_1(x))\cdots))}.$$

Se aplican **de dentro hacia fuera**, empezando por $f_1$. La composición es **asociativa** cuando las operaciones están definidas: podemos agruparlas sin alterar el resultado, pero **no cambiar su orden**.

Ejemplo con tres pasos:

$$f_1(x)=x+2,\quad f_2(x)=x^2,\quad f_3(x)=\sqrt{x}.$$

$$x\longrightarrow x+2\longrightarrow(x+2)^2\longrightarrow\sqrt{(x+2)^2}=|x+2|.$$

Así, $F(x)=|x+2|$ y $D_F=\mathbb R$. La raíz cuadrada devuelve el valor **no negativo**: $\sqrt{a^2}=|a|$, no siempre $a$.

```python
dibujar([(x,x+2,'Paso 1: x+2'),(x,(x+2)**2,'Paso 2: (x+2)²'),(x,np.abs(x+2),'Paso 3: |x+2|')], xlim=(-5,3), ylim=(-2,14), titulo='Tres funciones aplicadas sucesivamente')
```

### 3.2 Dominio de una composición: condición general

No basta con que $x$ esté en el dominio de la primera función: **cada resultado intermedio debe ser válido para la siguiente**.

Para dos funciones:

$$\boxed{D_{f\circ g}=\{x\in D_g:\ g(x)\in D_f\}}.$$

Para $n$ funciones, se exige que $x\in D_{f_1}$, que $f_1(x)\in D_{f_2}$, que $f_2(f_1(x))\in D_{f_3}$, y así sucesivamente.

**Ejemplo:** $f(x)=\sqrt{x}$, $g(x)=x-3$:

$$f(g(x))=\sqrt{x-3},\qquad D_{f\circ g}=[3,+\infty).$$

En cambio, $g(f(x))=\sqrt{x}-3$, cuyo dominio es $[0,+\infty)$.

La composición no siempre es posible para todas las entradas: depende de que las salidas de una función sean entradas admisibles para la siguiente.

```python
a=np.linspace(3,9,500); c=np.linspace(0,9,500)
dibujar([(a,np.sqrt(a-3),'f∘g = √(x−3)'),(c,np.sqrt(c)-3,'g∘f = √x−3')], xlim=(-1,9), ylim=(-4,4), titulo='El orden también cambia el dominio')
```

## 4. Función inversa

Una función inversa **deshace** la transformación de la función original. Se escribe $f^{-1}$; **no** significa $1/f(x)$.

Si $f(2)=7$, entonces $f^{-1}(7)=2$, siempre que exista esa inversa.

### 4.1 ¿Cuándo existe?

- Para que una función tenga inversa **sobre su imagen**, debe ser **inyectiva**: entradas diferentes producen salidas diferentes.
- Si queremos una inversa definida en **todo el codominio declarado**, la función debe ser **biyectiva**: inyectiva y sobreyectiva.
- Una función estrictamente creciente o estrictamente decreciente en un intervalo es inyectiva en ese intervalo.
- **Prueba gráfica:** toda recta horizontal debe cortar la gráfica como máximo una vez para que sea inyectiva.

Ejemplo: $f(x)=x^2$ no es inyectiva en $\mathbb R$ porque $f(2)=f(-2)=4$. Si restringimos su dominio a $[0,+\infty)$, sí lo es y su inversa es $\sqrt{x}$. También podríamos restringirlo a $(-\infty,0]$, obteniendo como inversa $-\sqrt{x}$.

```python
a=np.linspace(0,5,500)
dibujar([(a,a*a,'f(x) = x², x ≥ 0'),(a,np.sqrt(a),'f⁻¹(x) = √x')], xlim=(-.5,5), ylim=(-.5,7), diagonal=True, titulo='Función e inversa: simetría respecto de y = x')
```

### 4.2 Cómo calcular la inversa

Ejemplo: $f(x)=2x+3$.

1. Escribimos $y=2x+3$.
2. Intercambiamos $x$ e $y$: $x=2y+3$.
3. Despejamos $y$: $y=(x-3)/2$.
4. Por tanto, $\boxed{f^{-1}(x)=(x-3)/2}$.

**Importante:** intercambiar letras y despejar no garantiza que exista una función inversa: primero hay que comprobar la inyectividad o restringir el dominio.

### 4.3 Qué deben cumplir una función y su inversa

Si $f:A\to B$ es biyectiva y $f^{-1}:B\to A$ es su inversa, entonces:

$$\boxed{f^{-1}(f(x))=x\quad\text{para todo }x\in A}$$

$$\boxed{f(f^{-1}(y))=y\quad\text{para todo }y\in B}$$

Además:

- $D_{f^{-1}}=\operatorname{Im}(f)$ y $\operatorname{Im}(f^{-1})=D_f$.
- Sus gráficas son simétricas respecto a la recta $y=x$.
- Si $(a,b)$ pertenece a la gráfica de $f$, entonces $(b,a)$ pertenece a la de $f^{-1}$.

**Las dos composiciones deben dar la identidad en sus respectivos dominios.**

```python
a=np.linspace(-4,4,500)
dibujar([(a,2*a+3,'f(x) = 2x+3'),(a,(a-3)/2,'f⁻¹(x) = (x−3)/2')], xlim=(-4,4), ylim=(-5,7), diagonal=True, titulo='Recta y su inversa')
```

### 4.4 Composición e inversa de varias funciones

Si $F=f_n\circ\cdots\circ f_2\circ f_1$ y todas las funciones implicadas tienen las inversas adecuadas, entonces:

$$\boxed{F^{-1}=f_1^{-1}\circ f_2^{-1}\circ\cdots\circ f_n^{-1}}.$$

**Para deshacer una cadena, se invierte el orden.** Si primero sumamos 2, después multiplicamos por 3 y finalmente elevamos al cubo, para regresar hay que sacar raíz cúbica, dividir entre 3 y restar 2.

Ejemplo:

$$F(x)=[3(x+2)]^3,\qquad F^{-1}(x)=\frac{\sqrt[3]{x}}{3}-2.$$

Comprueba sustituyendo que ambas composiciones devuelven $x$.

```python
a=np.linspace(-4,1,600); y=(3*(a+2))**3
z=np.linspace(-60,60,700)
dibujar([(a,y,'F(x) = [3(x+2)]³'),(z,np.cbrt(z)/3-2,'F⁻¹(x) = ∛x/3−2')], xlim=(-5,5), ylim=(-12,12), diagonal=True, titulo='Inversa de una composición de tres funciones')
```

## 5. Errores frecuentes y método de examen

1. **Dominio no es resolver $f(x)=0$:** buscamos dónde tiene sentido evaluar la función.
2. **Raíz par:** radicando $\ge0$; si está en el denominador, $>0$.
3. **Raíz impar:** admite negativos, pero no anula otras restricciones.
4. **Logaritmo:** argumento estrictamente positivo.
5. **Varias restricciones:** tomar la **intersección**.
6. **Composición:** calcular de dentro hacia fuera y comprobar cada dominio intermedio.
7. **Inversa:** comprobar inyectividad; $f^{-1}(x)$ no es $1/f(x)$.
8. **Identidades de inversas:** respetar los dominios de cada composición.
