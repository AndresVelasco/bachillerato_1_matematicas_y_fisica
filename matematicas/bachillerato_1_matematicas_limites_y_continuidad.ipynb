## 1. ¿Qué es un límite?

El límite describe **a qué valor se acerca** $f(x)$ cuando $x$ se aproxima a un punto, **no necesariamente cuánto vale** la función en ese punto.

$$\lim_{x\to a}f(x)=L$$

- $x\to a^-$: nos acercamos a $a$ **por la izquierda**.
- $x\to a^+$: nos acercamos a $a$ **por la derecha**.
- Existe un límite finito en $a$ **si y solo si** los dos límites laterales existen, son finitos y coinciden.

**Ejemplo: un agujero en la gráfica.** Para $x\neq2$,

$$f(x)=\frac{x^2-4}{x-2}=x+2.$$

Por tanto, $\lim_{x\to2}f(x)=4$, aunque $f(2)$ no está definida. El círculo vacío señala ese punto excluido.

```python
x = np.linspace(-1, 5, 300)
plt.figure(figsize=(7, 4))
plt.plot(x, x + 2)
plt.scatter([2], [4], s=90, facecolors='white', edgecolors='C0', linewidths=2, zorder=5)
plt.axvline(2, linestyle='--', color='gray', alpha=.5)
plt.title('Límite 4 aunque la función no existe en x = 2')
plt.grid(alpha=.25)
plt.show()
```

### Límites laterales diferentes

Si $\lim_{x\to a^-}f(x)\ne\lim_{x\to a^+}f(x)$, **no existe el límite bilateral**. Por ejemplo, si $f(x)=1$ para $x<0$ y $f(x)=3$ para $x\ge0$, los límites laterales son 1 y 3.

```python
plt.figure(figsize=(7, 4))
plt.plot([-3, 0], [1, 1]); plt.plot([0, 3], [3, 3])
plt.scatter([0], [1], facecolors='white', edgecolors='C0', s=90, linewidths=2)
plt.scatter([0], [3], color='C1', s=55)
plt.xlim(-3, 3); plt.ylim(0, 4); plt.grid(alpha=.25)
plt.title('Salto: los límites laterales no coinciden'); plt.show()
```

## 2. Tipos de límites

| Situación | Ejemplo | Interpretación |
|---|---|---|
| Límite finito en un punto | $\lim_{x\to2}(x+1)=3$ | Se aproxima a 3 |
| Límite infinito en un punto | $\lim_{x\to0^+}1/x=+\infty$ | Crece sin cota al acercarse a 0 por la derecha |
| Límite en el infinito | $\lim_{x\to+\infty}1/x=0$ | Se aproxima a 0 cuando $x$ crece |

**Atención:** escribir $+\infty$ o $-\infty$ no significa que el límite sea un número real. Es una descripción del comportamiento de la función.

### Signos al acercarse a cero

Para $1/x$, el límite por la derecha de cero es $+\infty$ y por la izquierda es $-\infty$. Para $1/x^2$, ambos laterales son $+\infty$. Por ello hay que estudiar los **signos**, no solo sustituir.

```python
x1 = np.linspace(-3, -.08, 400)
x2 = np.linspace(.08, 3, 400)
plt.figure(figsize=(7, 4))
for x in (x1, x2): plt.plot(x, 1/x, color='C0')
plt.axvline(0, linestyle='--', color='C1'); plt.axhline(0, color='gray', linewidth=.7)
plt.ylim(-8, 8); plt.grid(alpha=.25); plt.title('1/x: signos opuestos a ambos lados de 0'); plt.show()
```

## 3. Propiedades y cálculo directo

Si $\lim f(x)=A$ y $\lim g(x)=B$ (límites finitos en el mismo punto), entonces:

$$\lim(f\pm g)=A\pm B,\qquad \lim(fg)=AB,\qquad \lim\frac fg=\frac AB\quad(B\ne0).$$

También se pueden calcular potencias, y raíces cuando estén definidas y se cumplan las condiciones de continuidad correspondientes.

**Sustitución directa:** si la función es continua en $a$, basta calcular $f(a)$. Ejemplo: $\lim_{x\to3}(x^2-2x)=9-6=3$.

**Composición:** si $g(x)\to b$ y $f$ es continua en $b$, entonces $f(g(x))\to f(b)$. No se debe sustituir sin comprobar las condiciones.

## 4. Indeterminaciones

Una **indeterminación** no es un resultado: indica que hay que transformar la expresión. No hay que confundir $0/0$ con $k/0$ ($k\ne0$): en este último caso se estudian los signos y puede aparecer un límite infinito.

| Forma | Técnica habitual | Ejemplo |
|---|---|---|
| $0/0$ | Factorizar y simplificar | $\frac{x^2-4}{x-2}$, cuando $x\to2$ |
| $0/0$ con raíces | Multiplicar por el conjugado | $\frac{\sqrt{x+1}-1}{x}$, cuando $x\to0$ |
| $\infty/\infty$ | Dividir por la mayor potencia relevante | $\frac{2x^2+1}{x^2-3}$ |
| $\infty-\infty$ | Denominador común o racionalizar | $\sqrt{x^2+x}-x$ cuando $x\to+\infty$ |
| $0\cdot\infty$ | Reescribir como cociente | $x\cdot(1/x)$ cuando $x\to0$, $x\ne0$ |

### Ejemplos resueltos

**Factorización:**
$$\lim_{x\to2}\frac{x^2-4}{x-2}=\lim_{x\to2}\frac{(x-2)(x+2)}{x-2}=4.$$

**Racionalización:**
$$\lim_{x\to0}\frac{\sqrt{1+x}-1}{x}=\lim_{x\to0}\frac{1}{\sqrt{1+x}+1}=\frac12.$$

**Diferencia de infinitos:** para $x\to+\infty$,
$$\sqrt{x^2+x}-x=\frac{x}{\sqrt{x^2+x}+x}=\frac{1}{\sqrt{1+1/x}+1}\to\frac12.$$

**Cuidado:** $0\cdot\infty$ puede producir distintos resultados; el ejemplo $x(1/x)=1$ solo demuestra uno de ellos.

## 5. Límites de racionales en el infinito

Sea $f(x)=P(x)/Q(x)$, con $P,Q$ polinomios.

| Grados | Comportamiento al ir $x\to\pm\infty$ |
|---|---|
| $\deg P<\deg Q$ | Límite $0$ |
| $\deg P=\deg Q$ | Cociente de coeficientes principales |
| $\deg P>\deg Q$ | El módulo crece sin cota; **hay que determinar el signo** |

Ejemplos:

$$\lim_{x\to\infty}\frac{3x+1}{x^2+2}=0,\qquad
\lim_{x\to\infty}\frac{2x^2-1}{5x^2+3}=\frac25.$$

$$\lim_{x\to+\infty}\frac{x^3+1}{x^2+1}=+\infty,\qquad
\lim_{x\to-\infty}\frac{x^3+1}{x^2+1}=-\infty.$$

**Clave:** en $x\to-\infty$, una potencia par es positiva y una impar conserva el signo negativo.

## 6. Asíntotas

Una asíntota es una recta a la que la gráfica se aproxima de una manera precisa. **La gráfica puede cortar una asíntota horizontal u oblicua**.

### 6.1 Vertical: $x=a$

Existe si al menos uno de los límites laterales $\lim_{x\to a^-}f(x)$ o $\lim_{x\to a^+}f(x)$ es $+\infty$ o $-\infty$.

Ejemplo: $f(x)=1/(x-1)$ tiene asíntota vertical $x=1$.

### 6.2 Horizontal: $y=L$

Si $\lim_{x\to+\infty}f(x)=L$ **o** $\lim_{x\to-\infty}f(x)=L$, hay asíntota horizontal $y=L$ en ese extremo. Puede haber distintas asíntotas a derecha e izquierda.

Ejemplo: $f(x)=2+1/x$ tiene horizontal $y=2$.

### 6.3 Oblicua: $y=mx+n$, con $m\ne0$

En el extremo que se esté estudiando, si existen y son finitos:

$$m=\lim_{x\to\pm\infty}\frac{f(x)}x,\qquad n=\lim_{x\to\pm\infty}(f(x)-mx),$$

con $m\ne0$, entonces $y=mx+n$ es asíntota oblicua en ese extremo.

Ejemplo: $f(x)=x+1+1/x$ tiene oblicua $y=x+1$. En funciones racionales, la división de polinomios suele facilitar su cálculo.

**Procedimiento:** primero verticales (puntos problemáticos del dominio), luego horizontales en $+\infty$ y $-\infty$ y, si procede, oblicuas. En un mismo extremo no hay horizontal y oblicua a la vez.

```python
fig, ax = plt.subplots(1, 3, figsize=(13, 3.6))
for x in (np.linspace(-4, .9, 300), np.linspace(1.1, 5, 300)):
    ax[0].plot(x, 1/(x-1), color='C0')
ax[0].axvline(1, linestyle='--', color='C1'); ax[0].set_ylim(-8, 8); ax[0].set_title('Vertical: x = 1')
x = np.linspace(.2, 9, 300)
ax[1].plot(x, 2+1/x); ax[1].axhline(2, linestyle='--', color='C1'); ax[1].set_ylim(0, 8); ax[1].set_title('Horizontal: y = 2')
x = np.linspace(1, 8, 300)
ax[2].plot(x, x+1+1/x); ax[2].plot(x, x+1, '--', color='C1'); ax[2].set_title('Oblicua: y = x + 1')
for a in ax: a.grid(alpha=.25)
plt.tight_layout(); plt.show()
```

## 7. Continuidad

Una función es **continua en $x=a$** cuando se cumplen las tres condiciones:

1. $f(a)$ está definida.
2. Existe el límite finito $\lim_{x\to a}f(x)$.
3. $\boxed{\lim_{x\to a}f(x)=f(a)}$.

La intuición de «dibujar sin levantar el lápiz» ayuda, pero la definición matemática son estas condiciones. En los extremos de un intervalo se usa la continuidad lateral correspondiente.

**Funciones habituales:** los polinomios son continuos en $\mathbb R$; los cocientes de polinomios, donde el denominador no se anula; las raíces pares, donde su radicando es no negativo; los logaritmos, donde su argumento es positivo.

### Tipos de discontinuidad

- **Evitable:** existe límite finito, pero $f(a)$ no existe o no coincide con él. Se puede reparar asignando $f(a)$ igual al límite.
- **De salto finito:** los límites laterales son finitos pero distintos.
- **Infinita:** al menos un límite lateral es infinito (aparece una asíntota vertical).

Hay otras discontinuidades más complejas; no son necesarias aquí.

### Ejemplo visual: discontinuidad evitable

La curva siguiente tendría continuidad si el punto marcado se situara sobre el agujero.

```python
x = np.linspace(-2, 2, 300)
plt.figure(figsize=(7, 4)); plt.plot(x, x**2)
plt.scatter([1], [1], facecolors='white', edgecolors='C0', s=100, linewidths=2)
plt.scatter([1], [3], color='C1', s=65)
plt.annotate('f(1) = 3', (1, 3), xytext=(1.2, 3.4))
plt.grid(alpha=.25); plt.title('Límite 1, pero valor de la función 3'); plt.show()
```

## 8. Continuidad en funciones a trozos

En el punto donde cambia la fórmula hay que comprobar **ambos límites laterales y el valor de la función**.

$$f(x)=\begin{cases}x^2+1,&x<2\\ ax+1,&x\ge2\end{cases}$$

Para que sea continua en $x=2$:

$$\lim_{x\to2^-}f(x)=2^2+1=5,\qquad \lim_{x\to2^+}f(x)=2a+1=f(2).$$

Igualando $5=2a+1$, obtenemos **$a=2$**.

**Regla práctica:** nunca basta con igualar expresiones sin comprobar qué tramo incluye el punto y qué límites laterales corresponden.

```python
x_izq = np.linspace(-1, 2, 250, endpoint=False)
x_der = np.linspace(2, 4, 250)
plt.figure(figsize=(7, 4))
plt.plot(x_izq, x_izq**2+1, label='x² + 1, x < 2')
plt.plot(x_der, 2*x_der+1, label='2x + 1, x ≥ 2')
plt.scatter([2], [5], color='C1', zorder=5)
plt.grid(alpha=.25); plt.legend(); plt.title('Con a = 2, los tramos se unen sin salto'); plt.show()
```

## 9. Teorema de Bolzano

**Enunciado.** Si una función $f$ es continua en todo el intervalo cerrado $[a,b]$ y $f(a)$ y $f(b)$ tienen **signos opuestos**, entonces existe al menos un $c\in(a,b)$ tal que $f(c)=0$.

$$f(a)\,f(b)<0\quad\Longrightarrow\quad \exists c\in(a,b):f(c)=0,$$

**siempre que $f$ sea continua en $[a,b]$.**

### Ejemplo: demostrar que hay una raíz

Para $f(x)=x^3-x-1$:

- Es un polinomio, así que es continua en $[1,2]$.
- $f(1)=-1<0$.
- $f(2)=5>0$.

**Conclusión:** existe al menos una raíz en $(1,2)$.

Bolzano **no dice dónde está exactamente** la raíz ni que sea única. Tampoco garantiza que haya cambio de signo si existe una raíz (por ejemplo, $x^2$ toca el eje en cero sin cambiar de signo).

```python
x = np.linspace(0.8, 2.1, 400)
y = x**3 - x - 1
plt.figure(figsize=(7, 4)); plt.plot(x, y)
plt.axhline(0, color='gray', linewidth=1)
plt.scatter([1, 2], [-1, 5], color='C1', zorder=5)
plt.axvspan(1, 2, color='gray', alpha=.08)
plt.grid(alpha=.25); plt.title('Bolzano: continuidad y cambio de signo en [1, 2]'); plt.show()
```

### Aproximación por bisección (sin programar)

La idea consiste en partir el intervalo por la mitad y conservar el subintervalo donde hay cambio de signo:

1. Partimos de $[1,2]$.
2. En $1.5$, $f(1.5)=0.875>0$; la raíz está en $(1,1.5)$.
3. En $1.25$, $f(1.25)=-0.296875<0$; la raíz está en $(1.25,1.5)$.
4. En $1.375$, $f(1.375)=0.224609375>0$; la raíz está en $(1.25,1.375)$.

Repitiendo el procedimiento, obtenemos aproximaciones cada vez más precisas.

### ¿Por qué importan las hipótesis?

La función $f(x)=-1$ si $x<0$ y $f(x)=1$ si $x\ge0$ cambia de signo entre $-1$ y $1$, **pero no tiene raíces**: no es continua. Y $f(x)=x^2$ tiene una raíz en $0$ sin que los valores en $-1$ y $1$ tengan signos opuestos.

## 10. Cheatsheet de examen

| Pregunta | Qué hacer |
|---|---|
| ¿Límite en $a$? | Sustituir si es posible; si hay problema, estudiar laterales e indeterminaciones |
| ¿Sale $0/0$? | Factorizar, simplificar o racionalizar |
| ¿Sale $\infty/\infty$? | Comparar grados o dividir por potencia dominante |
| ¿Sale $\infty-\infty$? | Unificar fracciones o racionalizar |
| ¿Límite infinito? | Comprobar signo a cada lado |
| ¿Asíntota vertical? | Buscar puntos con límites laterales infinitos |
| ¿Horizontal? | Calcular límites en $+\infty$ y $-\infty$ |
| ¿Oblicua? | Calcular $m$ y $n$ en cada extremo, con $m\ne0$ |
| ¿Continuidad en $a$? | Verificar $f(a)$, límite finito y su igualdad |
| ¿Función a trozos? | Estudiar ambos laterales y el valor en el punto de unión |
| ¿Aplicar Bolzano? | Justificar continuidad, comprobar signos opuestos y concluir existencia |
